```python
df["Revenue"]=df["Quantity"]*df["Price"]*(1-df["Discount"])
```

```python
df["Cost"]=df["Revenue"]*0.70
df["Profit"]=df["Revenue"]-df["Cost"]
```

```python
df["Month"]=df["Date"].dt.to_period("M").astype(str)
monthly_sales=df.groupby("Month")["Revenue"].sum()
df["Monthly_Sales"]=df["Month"].map(monthly_sales)
```

```python
monthly_orders=df.groupby("Month")["Order ID"].nunique()
df["Monthly_Orders"]=df["Month"].map(monthly_orders)
```

```python
df["AOV"]=df["Monthly_Sales"]/df["Monthly_Orders"]
```

```python
mom=monthly_sales.pct_change()*100
df["MoM Sales %"]=df["Month"].map(mom)
df["MoM Sales %"]=df["MoM Sales %"].clip(lower=0)
```

```python
df.to_csv('superstore_sales.csv',index=False)
```

```python
df.head()
```

```python
product_sales=(
    df.groupby("Product")["Revenue"].sum().sort_values(ascending=False)
)
top_5_product=product_sales.head(5)
bottom_5_product=product_sales.tail(5)
```

```python
top_5_product.to_csv('top_5_product.csv')
```

```python
bottom_5_product.to_csv('bottom_5_product.csv')
```

```python
region_by_sales=(df.groupby("Region")["Revenue"].sum().sort_values(ascending=False))
region_by_sales
```

```python
top_5_regions=region_by_sales.head(5)
bottom_5_regions=region_by_sales.tail(5)
```

```python
top_5_regions.to_csv('top_5_regions.csv')
```

```python
bottom_5_regions.to_csv('bottom_5_regions.csv')
```

```python
df["Discount_group"] = pd.cut(df["Discount"], bins=[0, 0.2, 0.4, 0.6, 0.8, 1], labels=["0-20%", "20-40%", "40-60%", "60-80%", "80-100%"])
result = df.groupby("Discount_group", observed=False).agg(
    Sales=("Revenue", "sum"),
    Profit=("Profit", "sum")
).reset_index()
result.to_csv('discount analysis.csv', index=False)
result
```
